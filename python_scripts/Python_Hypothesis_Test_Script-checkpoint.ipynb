```python
import pandas as pd
import matplotlib.pyplot as plt
import scikit_posthocs as sp
from scipy.stats import kruskal

print("success")
```

```python
df = pd.read_csv("export_table_dataset.csv")

print(df)
print(df["postcode_area"].value_counts())
```

```python
summary = df.groupby("postcode_area")["normalised_price"].agg(["count", "mean", "median", "std", "min", "max"]) 

print(summary)
```

```python
summary.to_excel("summary_stats.xlsx")
```

```python
postcode = "EH4"

prices = df[
    df["postcode_area"] == postcode
]["normalised_price"].dropna()

plt.hist(prices, bins=10)

plt.xlabel("House Price")
plt.ylabel("Frequency")
plt.title(f"House Prices in {postcode}")

plt.show()
```

```python
postcode_areas = df["postcode_area"].dropna().unique()

# Create one subplot for each postcode area
fig, axes = plt.subplots(
    nrows=len(postcode_areas),
    ncols=1,
    figsize=(10, 5 * len(postcode_areas))
)

for ax, area in zip(axes, postcode_areas):

    prices = df[
        df["postcode_area"] == area
    ]["normalised_price"].dropna()

    if len(prices) >= 1:

        ax.hist(prices, bins=10)

        ax.set_xlabel("House Price")
        ax.set_ylabel("Frequency")
        ax.set_title(f"House Prices - {area}")

plt.tight_layout()
plt.show()
```

```python
area_counts = df["postcode_area"].value_counts()

valid_areas = area_counts[
    area_counts >= 30
].index

df_test = df[
    df["postcode_area"].isin(valid_areas)
].copy()

print("\nPostcode areas included in statistical tests:")
print(df_test["postcode_area"].value_counts().sort_index())
```

```python
groups = []

for area, group in df_test.groupby("postcode_area"):

    prices = group["normalised_price"].dropna()

    groups.append(prices)

    print(area, len(prices))
```

```python
#for area, group in df.groupby("postcode_area"):

#    prices = group["price"].dropna()

#    if len(prices) >= 10:
#        print(area, len(prices))
```

```python
statistic, p_value = kruskal(*groups)

print("Kruskal-Wallis Test")
print("-------------------")
print("Statistic:", statistic)
print("P-value:", p_value)
```

```python
posthoc = sp.posthoc_dunn(
    df_test,
    val_col="normalised_price",
    group_col="postcode_area",
    p_adjust="bonferroni"
)

print("\nDunn's Test:")
print(posthoc)
```
